```python
from pyspark.sql import functions as f , types as t

data = [
    (100, 'Avishek', 24, "Raniganj", "123xxx"),
    (101, 'Tuhina', 27, "Nadia", "123asy"),
    (102, 'Kalyan', 27, "Purulia", "124jbdk"),
    (104, 'Ram', 28, "Asansol", "123iqi"),
    (105, 'Shyam', 22, "Kolkata", "125dii"),
    (106, 'Nayan', 30, "Bengalore", "128ooo"),
]

schema = "id int, name string, age int, city string, Special_Id string"
df = spark.createDataFrame(data, schema)
df.show()
df = df.withColumn("Special_Id", f.regexp_replace("Special_Id", "123", "ABC"))
df.show()

```

```python
from pyspark.sql import functions as f , types as t

# Customer data with nulls
customers_data = [
    (100, 'Avishek', 24, "Raniganj", "9876543210"),
    (101, 'Tuhina', 27, "Nadia", None),  # null phone
    (102, 'Kalyan', None, "Purulia", "9123456789"),  # null age
    (103, 'Ram', 28, None, "9988776655"),  # null city
    (104, 'Shyam', 22, "Kolkata", "9876512345"),
    (105, 'Nayan', 30, "Bengalore", None),  # null phone
    (106, 'Priya', None, None, "9823456789"),  # null age and city
]

# Orders data with nulls
orders_data = [
    (1, 100, 'Pizza', 300.0, "2026-06-20"),
    (2, 101, 'Burger', 160.0, None),  # null date
    (3, None, 'Fries', 60.0, "2026-06-21"),  # null customer_id
    (4, 103, 'Cake', None, "2026-06-22"),  # null price
    (5, 104, 'Mango Shake', 80.0, "2026-06-23"),
    (6, 105, None, 250.0, "2026-06-24"),  # null product
    (7, None, 'Ice Cream', None, None),  # multiple nulls
]

# Create DataFrames
customers_schema = "customer_id int, name string, age int, city string, phone string"
orders_schema = "order_id int, customer_id int, product string, price double, order_date string"

df_customers = spark.createDataFrame(customers_data, customers_schema)
df_orders = spark.createDataFrame(orders_data, orders_schema)

# Display both DataFrames
print("Customers Table:")
df_customers.show()

print("\nOrders Table:")
df_orders.show()
```

```python
# INNER JOIN - Only matching records
print("INNER JOIN:")
df_inner = df_customers.join(df_orders, "customer_id", "inner")
df_inner.show()

# LEFT JOIN - All customers, matching orders
print("\nLEFT JOIN (All Customers):")
df_left = df_customers.join(df_orders, "customer_id", "left")
df_left.show()

# RIGHT JOIN - All orders, matching customers
print("\nRIGHT JOIN (All Orders):")
df_right = df_customers.join(df_orders, "customer_id", "right")
df_right.show()

# FULL OUTER JOIN - All records from both tables
print("\nFULL OUTER JOIN:")
df_outer = df_customers.join(df_orders, "customer_id", "outer")
df_outer.show()
```
