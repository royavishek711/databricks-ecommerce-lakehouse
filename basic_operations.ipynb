### Column Filtering

```python
df_trimmed = df.select('imdb_rating','release_year','studio','revenue','budget')
# display(df_trimmed)
df_trimmed.show(3, truncate=False)
```

### Row Filtering

```python
df_row_filtered = df.filter((df['release_year']>2000) & (df['release_year']<2010))
display(df_row_filtered)
```

```python
from pyspark.sql.functions import col

df_marvel = df.filter(col('studio') == 'Marvel Studios')
display(df_marvel)
```

```python
unique_indistries = df.select("industry").distinct()
display(unique_indistries)
```

```python
unique_lang = df.select("language").distinct()
display(unique_lang)
```

```python
from pyspark.sql.functions import round

# to see the profit and store in same dataframe with new column, column must be within 2 decimal range
df = df.withColumn('Profit', round(col("revenue") - col("budget"), 2))
display(df)
```

```python
# Column Renaming 
df = df.withColumnRenamed("revenue", "Total_Revenue")
df.printSchema()
```
